```python
import geopandas as gpd
from matplotlib import pyplot as plt
import pandas as pd
```

```python
countries = pd.read_csv("/Users/patnaik/Documents/Yogi/My-Countries.csv")
exclude = ["Antarctica", "Cyprus U.N. Buffer Zone", "Northern Cyprus", "Zanzibar"]
countries = countries[~countries.iloc[:, 0].isin(exclude)]
countries = countries.reset_index(drop=True)
countries
```

```python
worldmap = gpd.read_file("/Users/patnaik/Documents/30DayMapChallenge/Day 3 - Polygons/ne_10m_admin_0_countries/ne_10m_admin_0_countries.shp")
```

```python
countries = countries.replace("Tanzania", "United Republic of Tanzania")
visited = worldmap[
    worldmap["ADMIN"].isin(countries.iloc[:, 0])
].copy()
visited
```

```python
fig, ax = plt.subplots(figsize=(12, 6))

# Background
fig.patch.set_facecolor("lightsteelblue")
ax.set_facecolor("powderblue")

# Leave Antarctica out of the display
background = worldmap[worldmap["ADMIN"] != "Antarctica"]

background.to_crs("ESRI:54030").plot(
    ax=ax,
    facecolor="floralwhite",
    edgecolor="gray",
    linewidth=0.3
)

visited.to_crs("ESRI:54030").plot(
    ax=ax,
    color="darkolivegreen",
    edgecolor="#101820",
    linewidth=0.3
)

ax.set_axis_off()
ax.set_title(
    "Countries I've Visited",
    fontname="Sinhala MN",
    fontsize=22,
    color="black",
    pad=20
)

# Credits
fig.text(
    0.13, 0.08,
    "Map: Yogi Patnaik · Boundaries: Natural Earth (1:10m)",
    fontsize=7,
    color="black"
)
plt.savefig("countriesvisited.png", bbox_inches= "tight",dpi=400)
plt.show()
```
